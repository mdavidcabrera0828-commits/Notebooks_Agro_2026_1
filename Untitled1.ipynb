```python
# Sistema de Gestión de Consumo Hídrico: "La Pluma de Agua"

# 1. ENTRADA DEL SISTEMA
print("====================================================")
print("   BIENVENIDO AL SISTEMA DE CONSUMO 'LA PLUMA'      ")
print("====================================================\n")

# Se solicita al usuario el número de plumas asignadas (puede ser un número decimal)
num_plumas = float(input("Ingrese el número de plumas asignadas al hogar: "))

# 2. PROCESO

# Equivalencia: 1 pluma = 0.025 litros por segundo
LITROS_POR_SEGUNDO_POR_PLUMA = 0.025

# A. Calcular el caudal total en litros por segundo
caudal_lps = num_plumas * LITROS_POR_SEGUNDO_POR_PLUMA

# B. Transformar el caudal a consumo diario (litros por día)
# 1 minuto = 60 segundos | 1 hora = 60 minutos | 1 día = 24 horas
# Segundos en un día = 60 * 60 * 24 = 86,400 segundos
segundos_en_dia = 86400
consumo_diario = caudal_lps * segundos_en_dia

# C. Transformar el consumo diario en consumo mensual (asumiendo un mes de 30 días)
consumo_mensual = consumo_diario * 30

# D. Clasificar el nivel de consumo según las reglas establecidas
if consumo_diario < 1000:
    clasificacion = "Bajo"
elif 1000 <= consumo_diario <= 3000:
    clasificacion = "Medio"
else:
    clasificacion = "Alto"

# 3. SALIDA
print("\n====================================================")
print("                RESUMEN DE CONSUMO                  ")
print("====================================================")
print(f"• Caudal asignado      : {caudal_lps:.3f} litros/segundo")
print(f"• Consumo diario       : {consumo_diario:.2f} litros/día")
print(f"• Consumo mensual (30d): {consumo_mensual:.2f} litros/mes")
print(f"• Clasificación de uso : {clasificacion}")
print("====================================================")
```

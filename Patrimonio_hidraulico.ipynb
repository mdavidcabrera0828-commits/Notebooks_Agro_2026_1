#**Situacion Problema**

En la ciudad de Sincelejo se decide aplicar un nuevo sistema para la medición del agua que llega a los hogares, basándose en la herencia hidráulica colonial de la región, para esto  utiliza una antigua unidad de medida conocida como plumas de agua, cada hogar sincelejano recibe cierta cantidad e plumas de agua. La población necesita entender cuanta cantidad de agua llega y se consume en su casa en unidades ya conocidas (litros/ unidad de tiempo) por ende se busca un sistema que pueda convertir la cantidad de plumas recibidas y poder calcular cuanto se consume de este recurso.

Un hogar en especifico ubicado en el barrio de la ciudadela universitaria recibe 0,5 plumas de agua, este necesita saber cuanto se consume y cuanta cantidad en especifico del recurso recibe.


##**1. Algoritmo en Lenguaje Natural**

Este algoritmo describe de forma descriptiva y secuencial el funcionamiento del sistema de medición hídrica para Sincelejo:

**1 Inicio del programa:** Se despliega un mensaje de bienvenida al usuario.

**2 Entrada de datos:** El sistema solicita al usuario que ingrese de forma numérica la cantidad de plumas de agua asignadas a su vivienda (para el caso de estudio de la Ciudadela Universitaria, el usuario ingresará 0.5).

**3 Definición de constantes:**
- Se establece el valor técnico de una pluma de agua equivalente en caudal ($0.02546\text{ litros por segundo}$).

- Se define la constante de tiempo correspondiente a los segundos que transcurren en un día ($86400\text{ segundos}$).

**4 Fase de procesamiento y cálculos:**
- **Caudal total:** Se multiplica el número de plumas ingresado por el valor base del caudal de una pluma.
- **Consumo diario:** Se toma el caudal total y se multiplica por los segundos de un día para transformar la unidad a litros diarios.
- **Consumo mensual:** Se proyecta el consumo diario multiplicándolo por los 30 días del mes comercial.

**5 Evaluación de condiciones (Clasificación del consumo):**
- El sistema verifica si el consumo diario es menor a 1000 litros; de ser así, asigna la categoría de nivel "Bajo".
- Si la condición anterior no se cumple, verifica si el consumo diario se encuentra en el rango de 1000 a 3000 litros inclusive; de ser así, asigna el nivel "Medio".
- Si no se cumple ninguna de las anteriores, por descarte asigna el nivel "Alto".

**6 Salida de resultados:** Se imprimen en pantalla los valores calculados de caudal total, consumo diario, consumo mensual y la etiqueta del nivel determinado.

**7 Fin:** Termina la secuencia del algoritmo.

##Pseudocódigo:

ALGORITMO Pluma_De_Agua

INICIO

Mostrar "Sistema de medición hídrica - Plumas de Agua"

Leer numero_plumas

pluma_lps ← 0.02546
segundos_dia ← 86400
dias_mes ← 30

caudal_total ← numero_plumas * pluma_lps

consumo_diario ← caudal_total * segundos_dia

consumo_mensual ← consumo_diario * dias_mes

SI consumo_diario < 1000 ENTONCES
    clasificacion ← "Bajo"
SINO SI consumo_diario <= 3000 ENTONCES
    clasificacion ← "Medio"
SINO
    clasificacion ← "Alto"
FIN SI

Mostrar caudal_total
Mostrar consumo_diario
Mostrar consumo_mensual
Mostrar clasificacion

FIN

```python
def simular_sistema_agua():
    # 1. Definir constantes del sistema hidráulico
    CAUDAL_UNA_PLUMA = 0.025463  # Litros por segundo (l/s)
    SEGUNDOS_POR_DIA = 86400  # Segundos en un día (24h)

    print("=========================================================")
    print("      SISTEMA DE MEDICIÓN DE AGUA - SINCELEJO            ")
    print("=========================================================\n")

    # --- RESOLUCIÓN DEL CASO ESPECÍFICO (Ciudadela Universitaria) ---
    print(">>> ANALIZANDO CASO: HOGAR EN CIUDADELA UNIVERSITARIA <<<")
    num_plumas = 0.5  # Dato de la situación problema

    # Calcular caudal total, consumo diario y mensual
    caudal_total = num_plumas * CAUDAL_UNA_PLUMA
    consumo_diario = caudal_total * SEGUNDOS_POR_DIA
    consumo_mensual = consumo_diario * 30

    # Determinar el nivel según las condiciones de la guía
    if consumo_diario < 1000:
        nivel = "Bajo"
    elif 1000 <= consumo_diario <= 3000:
        nivel = "Medio"
    else:
        nivel = "Alto"

    # Mostrar resultados del caso de estudio
    print(f"Cantidad de plumas asignadas: {num_plumas}")
    print(f"Caudal total: {caudal_total:.4f} litros/segundo")
    print(f"Consumo diario: {consumo_diario:.2f} litros/día")
    print(f"Consumo mensual: {consumo_mensual:.2f} litros/mes")
    print(f"Nivel de consumo: {nivel}")
    print("---------------------------------------------------------\n")

    # --- MODO INTERACTIVO GENERAL ---
    print(">>> CALCULADORA GENERAL PARA OTROS HOGARES <<<")
    try:
        plumas_usuario = float(
            input("Ingrese el número de plumas del hogar a evaluar: ")
        )

        # Cálculos para el usuario
        caudal_user = plumas_usuario * CAUDAL_UNA_PLUMA
        diario_user = caudal_user * SEGUNDOS_POR_DIA
        mensual_user = diario_user * 30

        # Clasificación del nivel
        if diario_user < 1000:
            nivel_user = "Bajo"
        elif 1000 <= diario_user <= 3000:
            nivel_user = "Medio"
        else:
            nivel_user = "Alto"

        # Mostrar resultados del usuario
        print(f"\nResultados:")
        print(f"• Caudal Total: {caudal_user:.4f} l/s")
        print(f"• Consumo Diario: {diario_user:.2f} litros/día")
        print(f"• Consumo Mensual: {mensual_user:.2f} litros/mes")
        print(f"• Nivel: {nivel_user}")

    except ValueError:
        print("\n[Error]: Ingrese un valor numérico válido.")


# Ejecución del programa
if __name__ == "__main__":
    simular_sistema_agua()
```
